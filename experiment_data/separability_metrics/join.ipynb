```python
import pandas as pd

gs_cc = pd.read_csv("geometric_separability_cc.csv")
gs_no_cc = pd.read_csv("geometric_separability_no_cc.csv")
gs_k = pd.read_csv("geometric_separability_k.csv")
```

```python
index = ["dataset", "split", "model", "layer", "epoch"]
gs_cc_idx = gs_cc.set_index(index)
gs_no_cc_idx = gs_no_cc.set_index(index)
gs_cc_idx.update(gs_no_cc_idx)

gs_cc_idx.reset_index().dropna(axis=1, how="all").to_csv("geometric_separability_geom.csv", index=False)
```

```python
gs_k = gs_k.dropna(axis=1, how="all")
gs_k_idx = gs_k.set_index(index)
gs_k_idx = gs_k_idx.iloc[:, 4:-2]
```

```python
gs_k_idx
```

```python
gs_full = pd.merge(gs_cc_idx.reset_index(), gs_k_idx.reset_index(), on=index, how="outer", suffixes=(None, ""))
```

```python
gs_full.to_csv("geometric_separability_full.csv", index=False)
```

### Merge Corrs

```python
cc_corrs = pd.read_csv("geometric_separability_cc_corrs.csv")
no_cc_corrs = pd.read_csv("geometric_separability_no_cc_corrs.csv")
k_corrs = pd.read_csv("geometric_separability_k_corrs.csv")

all_corrs = pd.concat([cc_corrs, no_cc_corrs, k_corrs], ignore_index=True).dropna()
all_corrs.rename(columns={"Unnamed: 0": "split"}, inplace=True)
all_corrs.to_csv("geometric_separability_all_corrs.csv", index=False)
```
