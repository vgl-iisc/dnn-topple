```python
import pickle
import pandas as pd

# file = "knn_times_evis.pkl" # data reported in eurovis submission
file = "./rn_times.pkl"
knn_times = {}
with open(file, "rb") as f:
    knn_times = pickle.load(f)
```

```python
keys_with_multiple = [k for k, v in knn_times.items() if len(v) > 1]
```

```python
print(file)
data = []

for k in keys_with_multiple:
    n, d = k
    times = knn_times[k]
    mean = sum(times) / len(times)
    max_dev = max(abs(mean - x) for x in times)
    std_dev = sum(abs(mean - x) for x in times) / (len(times) + 1)
    
    data.append({"Data Points": n, "Dimension": d, "Mean Time (s)": mean, "Max Deviation (s)": max_dev, "Std. Deviation (s)": std_dev})

df = pd.DataFrame(data)
df
```
