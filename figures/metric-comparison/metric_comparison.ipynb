```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import matplotlib.pyplot as plt

import plotly.io as pio
pio.renderers.default = "vscode"
pio.templates.default = "ggplot2"

color_seq = px.colors.qualitative.Dark24
color_seq_train = px.colors.qualitative.Pastel
color_seq_val = px.colors.qualitative.Set1

metric_data = pd.read_csv("../../experiment_data/separability_metrics/geometric_separability_subset_corrs_final.csv")
```

```python
metric_data.replace({"split": {"cifar_resnet": "ResNet", "cifar_densenet": "DenseNet", "full": "Combined"}, "metric": {"centroid_mean_separation": "Centroid Separation", "knn_accuracy_20": "KNN Accuracy", "silhouette_score": "Silhouette Score", "sackin_index_k20": "Sackin Index", "average_sackin_index_k20": "Avg Sackin", "centroid_mean_correlation": "Centroid Correlation", "total_persistence": "Total Persistence", "minima": "#Minima", "cps": "#Critical Points"}}, inplace=True)
```

```python
metric_data_f = metric_data[metric_data["metric"] != "Avg Sackin"]
```

```python
metric_order = [
	"KNN Accuracy",
	"Silhouette Score",
	"Centroid Separation",
	"Centroid Correlation",
	"Sackin Index",
    "Total Persistence",
    "#Minima",
    "#Critical Points"
]

metric_data_f["metric"] = pd.Categorical(
	metric_data_f["metric"],
	categories=metric_order,
	ordered=True
)

metric_data_f = metric_data_f.sort_values("metric")
metric_data_f
```

```python
fig = px.scatter(metric_data_f, x="value", y="metric", color="split", color_discrete_sequence=color_seq_val, )
fig.update_yaxes(title="", automargin=True, title_standoff=18)
fig.update_xaxes(title="Correlation with Validation Accuracy", title_standoff=18, automargin=True)
fig.update_layout(margin=dict(l=0, r=0, t=0, b=0), width=400*1.2*2, height=120*1.2*2, font=dict(size=16), legend=dict(
	xanchor="right", yanchor="top", x=0.98, y=0.98, bgcolor="rgba(255,255,255,0.5)", font=dict(size=15), title=""))
fig.update_traces(marker=dict(size=12))
fig.write_image(f"metrics.png", scale=4)
fig.show()
```
