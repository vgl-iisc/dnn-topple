```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import matplotlib.pyplot as plt

import plotly.io as pio
pio.renderers.default = "vscode"
pio.templates.default = "ggplot2"

color_seq = px.colors.qualitative.Dark24
color_seq_train = px.colors.qualitative.Pastel
color_seq_val = px.colors.qualitative.Set1
```

```python
c2f_26 = pd.read_csv("c2f_counts_26.csv")
```

```python
fig = make_subplots(rows=1, cols=1)
fig.add_trace(go.Bar(x=c2f_26["Class"], y=c2f_26["Count"], name="Contained in", marker=dict(color=color_seq[0])))
fig.add_trace(go.Bar(x=c2f_26["Class"], y=c2f_26["Count Majority"], name="Majority in", marker=dict(color=color_seq[1], opacity=0.8)))
fig.add_trace(go.Bar(x=c2f_26["Class"], y=c2f_26["Count Homogeneous"], name="Homogeneous in", marker=dict(color=color_seq[2], opacity=0.6)))
fig.update_layout(barmode='overlay')

fig.update_layout(margin=dict(l=0, r=0, t=0, b=0), width=400*2, height=180*2, font=dict(size=16))
fig.update_xaxes(title_text="Class", title_standoff=18, automargin=True)
fig.update_yaxes(title_text="#Valleys", title_standoff=18, automargin=True)
fig.write_image(f"plots/c2f_26.png", scale=8)
```

```python
c2f_1e4 = pd.read_csv("c2f_counts_1e-4.csv")
```

```python
a = 3
b = 0

fig = make_subplots(rows=1, cols=1)
fig.add_trace(go.Bar(x=c2f_1e4["Class"], y=c2f_1e4["Count"], name="Non-Homogeneous", marker=dict(color=color_seq[a])))
fig.add_trace(go.Bar(x=c2f_1e4["Class"], y=c2f_1e4["Count Majority"], name="Non-Homogeneous", marker=dict(color=color_seq[a]), showlegend=False))
fig.add_trace(go.Bar(x=c2f_1e4["Class"], y=c2f_1e4["Count Homogeneous"], name="Homogeneous", marker=dict(color=color_seq[b])))
fig.update_layout(barmode='overlay')

fig.update_layout(margin=dict(l=0, r=0, t=0, b=0), width=400*2, height=180*2, font=dict(size=16), legend=dict(
	xanchor="right", yanchor="top", x=0.98, y=0.98, bgcolor="rgba(255,255,255,0.5)", font=dict(size=14)))
fig.update_xaxes(title_text="Class", title_standoff=18, automargin=True)
fig.update_yaxes(title_text="#Valleys", title_standoff=18, automargin=True)
fig.write_image(f"plots/c2f_1e4.png", scale=8)
fig.show()
```
