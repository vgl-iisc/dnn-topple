```python
N = 250
C = 2

np.random.set_state(state)
centers = np.random.uniform(-5, 5, (C, 2))

# random covariance matrix
spreads = []
for _ in range(C):
	A = (np.random.rand(2, 2) - 0.5) * 3
	cov = np.dot(A, A.transpose())  # make it symmetric positive definite
	spreads.append(cov)

points = []
for i in range(C):
	pts = np.random.multivariate_normal(mean=centers[i], cov=spreads[i], size=N)
	points.append(pts)
 
points = np.vstack(points)
labels = np.array([i for i in range(C) for _ in range(N)])

fig = px.scatter(x=points[:, 0], y=points[:, 1], color=labels.astype(str), color_discrete_sequence=colors)
fig.update_layout(showlegend=False, width=400, height=400, margin=dict(l=0, r=0, t=0, b=0))
fig.update_xaxes(title_text=None, range=[-6, 6], showticklabels=False)
fig.update_yaxes(title_text=None, range=[-6, 6], showticklabels=False)
```

##### Loss-Augmented Surface

```python
import scipy.stats as stats

losses = []
for i in range(C):
	loss = stats.multivariate_normal.pdf(points[i*N:(i+1)*N], mean=centers[i], cov=spreads[i])
	losses.append(loss)

losses = np.concatenate(losses)

losses = 1.0 - (losses - np.min(losses)) / (np.max(losses) - np.min(losses)) * 3

support_pts = np.array([[-10.0, -10.0], [-10.0, 10.0], [10.0, -10.0], [10.0, 10.0]])

points_ext = np.vstack([points, support_pts])
max_loss = np.max(losses) * 1.25
losses_ext = np.hstack([losses, [max_loss, max_loss, max_loss, max_loss]])
```

```python
points_3d = np.hstack([points, losses.reshape(-1, 1)])
points_3d_ext = np.hstack([points_ext, losses_ext.reshape(-1, 1)])
```

```python
from scipy.spatial import Delaunay
import gpytoolbox as gpt

trig = Delaunay(points_ext)
V = points_3d_ext
F = trig.simplices
colors_ps = [(int(c[1:3], 16)/255, int(c[3:5], 16)/255, int(c[5:7], 16)/255) for c in colors]
```

```python
V_b, F_b, S = gpt.subdivide(V, F, method="loop", iters=3, return_matrix=True)

loss_b_ext = S @ losses_ext
```

```python
ps.init()
for i in range(C):
    mask = labels == i
    cloud = ps.register_point_cloud(f"Cluster {i}", points_3d[mask], color=colors_ps[i], radius=0.002)
    cloud.add_scalar_quantity("Loss", losses[mask])
    
surface = ps.register_surface_mesh("Loss Augmented Latents", V, F)
surface.add_scalar_quantity("Loss", V[:, 2])
surface = ps.register_surface_mesh("Loss Augmented Latents (smooth)", V_b, F_b)
surface.add_scalar_quantity("Loss", loss_b_ext)

ps.set_up_dir("z_up")
ps.show()
```

##### Split Tree

```python
import pyct as ct
from sklearn.neighbors import kneighbors_graph
import networkx as nx

graph = nx.from_scipy_sparse_array(kneighbors_graph(points, n_neighbors=20, mode='distance', include_self=False))

import utils as u

adjlist = u.get_adjlist_from_graph(graph)
scalar_field = losses

gsf = ct.GraphScalarFunction()

gsf.loadGraphFromAdjList(u.get_adjlist_from_graph(graph))
gsf.updateFnValues(list(scalar_field))

# Compute the contour tree
tree = ct.MergeTree()

tree.computeTree(gsf, ct.TreeType.TypeSplitTree)

outfile = "tree/ct"

tree.output(outfile, ct.TreeType.TypeSplitTree)
ctdata = ct.ContourTreeData()
ctdata.loadBinFile(outfile)

sim = ct.SimplifyCT()
sim.setInput(ctdata)

sim_fn = ct.Persistence(ctdata)

sim.simplify(sim_fn)
sim.outputOrder(outfile, False)
```
