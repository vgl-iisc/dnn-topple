```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import yaml

epoch_better_path = "../../experiment_data/balance_metrics/cross_epoch_better_e6.csv"

epoch_data = pd.read_csv(epoch_better_path)

import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import matplotlib.pyplot as plt

import plotly.io as pio
pio.renderers.default = "vscode"

color_seq = px.colors.qualitative.Dark24
color_seq_train = px.colors.qualitative.Pastel
color_seq_val = px.colors.qualitative.Set1
epoch_data.head()

window_size = 13
```

```python
epoch_trainUval = epoch_data[(epoch_data["split"] == "trainUval") & (epoch_data["model"] == "resnet") & (epoch_data["dataset"] == "cifar")]
epoch_trainUval = epoch_trainUval.sort_values(by='epoch')

epoch_trainUval["sackin_std"] = epoch_trainUval["sackin_index"].rolling(window=window_size).std()

fig = make_subplots(specs=[[{"secondary_y": True}]], )

train_acc = px.line(epoch_trainUval, x='epoch', y='train_acc', color="dataset", color_discrete_sequence=color_seq_train, symbol="model")
val_acc = px.line(epoch_trainUval, x='epoch', y='val_acc', color="dataset", color_discrete_sequence=color_seq_val, symbol="model")
sackin = px.scatter(epoch_trainUval, x='epoch', y='sackin_std', color="dataset", color_discrete_sequence=color_seq, symbol="model")

top_five_epochs = epoch_trainUval.nlargest(5, 'val_acc')["epoch"]

fig.add_trace(sackin.data[0], secondary_y=False)
fig.add_trace(train_acc.data[0], secondary_y=True)
fig.add_trace(val_acc.data[0], secondary_y=True)

for i, epoch in enumerate(top_five_epochs):
	gap = i * 3
	fig.add_vline(x=epoch, line_width=2, line_dash=f"5, {gap}", line_color=color_seq[i])
 
first_pctile = epoch_trainUval["sackin_std"].quantile(0.05)
fig.add_hline(y=first_pctile, line_width=2, line_dash="longdash", line_color="black")

# set title
fig.update_layout(
	margin=dict(l=0, r=0, t=0, b=0),
	showlegend=False,
 	width=600*2,
	height=300*2,
)

fig.update_xaxes(title_text="Epoch")
fig.update_yaxes(title_text="Previous 13 Sackin Std. Dev (log scale)", type="log", secondary_y=False)
fig.update_yaxes(title_text="Accuracy", secondary_y=True)
fig.show()
```
