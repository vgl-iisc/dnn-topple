```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import yaml

epoch_better_path = "../../experiment_data/balance_metrics/cross_epoch_better_e6.csv"

epoch_data = pd.read_csv(epoch_better_path)

import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import matplotlib.pyplot as plt

import plotly.io as pio
pio.renderers.default = "vscode"
pio.templates.default = "ggplot2"

color_seq = px.colors.qualitative.Dark24
color_seq_train = px.colors.qualitative.Pastel
color_seq_val = px.colors.qualitative.Set1
epoch_data.head()

window_size = 7
accuracy_window = 0.001
quant = 0.1
bg_color_dark = "rgba(0, 0, 0, 0.3)"
```

```python
epoch_trainUval = epoch_data[(epoch_data["split"] == "trainUval") & (epoch_data["model"] == "resnet") & (epoch_data["dataset"] == "cifar")]
epoch_trainUval = epoch_trainUval.sort_values(by='epoch')

epoch_trainUval["sackin_std"] = epoch_trainUval["sackin_index"].rolling(window=window_size).std()

fig = make_subplots(specs=[[{"secondary_y": True}]], )

train_acc = px.line(epoch_trainUval, x='epoch', y='train_acc', color="dataset", color_discrete_sequence=color_seq_train)
val_acc = px.line(epoch_trainUval, x='epoch', y='val_acc', color="dataset", color_discrete_sequence=color_seq_val)
sackin = px.scatter(epoch_trainUval, x='epoch', y='sackin_std', color="dataset", color_discrete_sequence=color_seq)

best_acc = epoch_trainUval["val_acc"].max()
epoch_trainUval["delta"] = best_acc - epoch_trainUval["val_acc"]

in_window = epoch_trainUval[epoch_trainUval["delta"] <= accuracy_window]
window = [in_window["epoch"].min(), in_window["epoch"].max()]

shape = dict(type="rect", xref="x", yref="paper", x0 = window[0], x1 = window[1], y0=0.0, y1=1.0, fillcolor=bg_color_dark, layer="below", line_width=0)

fig.add_trace(sackin.data[0], secondary_y=False)
fig.add_trace(train_acc.data[0], secondary_y=True)
fig.add_trace(val_acc.data[0], secondary_y=True)

first_pctile = epoch_trainUval["sackin_std"].quantile(quant)
fig.add_hline(y=first_pctile, line_width=2, line_dash="longdash", line_color="black")

fig.update_annotations(font_size=16)
fig.update_layout(margin=dict(l=0, r=0, t=0, b=0), width=400*2, height=180*2, font=dict(size=14), showlegend=False)
fig.update_xaxes(title_text="Epoch", title_standoff=18, automargin=True, range=[0,200])
fig.update_yaxes(title_text="Sackin Std. Dev", type="log", title_standoff=18, automargin=True, secondary_y=False)
fig.update_yaxes(title_text="Accuracy", title_standoff=18, automargin=True, secondary_y=True, nticks=10, range=[0, 1.05])
fig.add_shape(shape)

fig.write_image("resnet-cifar-stability.png", scale=8)
```
