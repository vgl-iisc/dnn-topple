```python
import pandas as pd
resnet = pd.read_csv(r"C:\home\eurovis_data\landscape_data_emnist_letters_last_best\resnet_emnist_letters\compiled_results.csv")
vgg = pd.read_csv(r"C:\home\eurovis_data\landscape_data_emnist_letters_last_best\vgg_emnist_letters\compiled_results.csv")
```

```python
def get_corrected_df(df):
    epoch = df["Epoch_No"].min()
    df = df[df["Epoch_No"] == epoch]
    df = df[df["Split"] == "Val"]
    
    return df
```

```python
resnet = get_corrected_df(resnet)
vgg = get_corrected_df(vgg)
```

```python
resnet_acc = resnet["Correct"].value_counts(normalize=True)[0]
vgg_acc = vgg["Correct"].value_counts(normalize=True)[0]
```
