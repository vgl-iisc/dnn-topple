```python
import pandas as pd
import plotly.express as px
from plotly.subplots import make_subplots

from glob import glob

kfiles = glob("../../experiment_data/balance_metrics/cross_epoch_k*_1e-6.csv")
kfiles = [(f, int(f.split("k")[1].split("_")[0])) for f in kfiles]

dfs = []
for f, k in kfiles:
	df = pd.read_csv(f)
	df["k"] = k
	dfs.append(df)

df = pd.concat(dfs, ignore_index=True)
df.sort_values(["model", "dataset", "layer", "k", "epoch"], inplace=True)
df.reset_index(drop=True, inplace=True)
df.to_csv("../../experiment_data/balance_metrics/cross_epoch_all_ks_1e-6.csv", index=False)
```

```python
varying_ks = df.groupby(["dataset", "model", "split", "layer", "epoch"])
```

```python
for (dataset, model, split, layer, epoch), group in varying_ks:
	print(group.iloc[:, 16])
	break
```

```python
kfiles_corr = glob("../../experiment_data/balance_metrics/cross_epoch_k*_1e-6_corr.csv")
kfiles_corr = [(f, int(f.split("k")[1].split("_")[0])) for f in kfiles_corr]

dfs = []
for f, k in kfiles_corr:
	df = pd.read_csv(f)
	df["k"] = k
	dfs.append(df)

df = pd.concat(dfs, ignore_index=True)
df.sort_values(["k"], inplace=True)
df.reset_index(drop=True, inplace=True)
df.rename(columns={"Unnamed: 0": "split"}, inplace=True)
df.to_csv("../../experiment_data/balance_metrics/cross_epoch_all_ks_1e-6_corr.csv", index=False)
df.head()
```
