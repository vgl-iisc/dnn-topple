```python
%load_ext autoreload
%autoreload 2
from experiment import find_all_datasets, find_all_experiments
import pyct as ct
```

```python
dses = find_all_datasets("../../datasets/")
imagenet = dses["imagenet"]
split = "trainUval"
```

```python
data_dir = "C:/home/eurovis_data/landscape_data_pt/"
strees_dir = "C:/home/eurovis_data/strees_pt/"
```

```python
exps = find_all_experiments(dses, data_dir, strees_dir)
exp = exps[1]
assert exp.split == split
```

```python
from basic_utils import get_labels, get_partition, get_tree, get_order_and_weights
labels = get_labels(exp)
partition = get_partition(exp)

assert len(partition) == len(labels)
```

```python
import numpy as np
import pickle as pkl

proportions = [0.9, 0.95, 0.99, 1.0]
res = {}

data, _ = get_tree(exp)
order, wts = get_order_and_weights(exp)

for prop in proportions:
	simpl = ct.SimplifyCT() # type: ignore
	simpl.setInput(data)
 
	res[prop] = simpl.getHomoValleyPlotPlusCoverages(order, wts, labels, prop, partition)
 
	print(f"Done with proportion {prop}")
```

```python
with open("imagenet_pt_homo_valley_plot_coverages.pkl", "wb") as f:
    pkl.dump(res, f)
```

```python
import os
from timeit import default_timer as dt
import pickle as pkl

basedir = "results"
os.makedirs(basedir, exist_ok=True)

proportions = [0.5, 0.6, 0.7, 0.8]

data, _ = get_tree(exp)
order, wts = get_order_and_weights(exp)

times = {}

for prop in proportions:
	simpl = ct.SimplifyCT() # type: ignore
	simpl.setInput(data)
 
	print(f"Starting computation of proportion {prop}")
	start = dt()
	rest = simpl.getHomoValleyPlotPlusCoverages(order, wts, labels, prop, partition)
	end = dt()
 
	times[prop] = end - start
	print(f"Time taken for proportion {prop}: {times[prop]} seconds")	
 
	with open(os.path.join(basedir, f"result_{prop}.pkl"), "wb") as f:
		pkl.dump(rest, f)
```
