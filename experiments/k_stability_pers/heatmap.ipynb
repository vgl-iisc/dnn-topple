```python
import pandas as pd

table = pd.read_csv("../../experiment_data/k_stability_pers/resnet_distances_e140.csv")
```

```python
table["k"] = list(table.columns)
```

```python
#  move column k to the front
table = table[["k"] + [col for col in table.columns if col != "k"]]
```

```python
table_f = table[table["k"] != "15"]
table_f.drop(columns=["15"], inplace=True)
# table_f.sort_values(by="k", ascending=False, inplace=True)
```

```python
import seaborn as sns
```

```python
fig = sns.heatmap(
	table_f.set_index("k"),
	annot=True,
	fmt=".3f",
	cmap="BuPu",
	vmin=0,
	vmax=12,
	cbar_kws={"label": "Distance"}
)

fig.xaxis.tick_top()
fig = fig.set_ylabel("").get_figure()
fig.savefig(f"plots/resnet_distances_e140.png", bbox_inches="tight", dpi=300)
```
