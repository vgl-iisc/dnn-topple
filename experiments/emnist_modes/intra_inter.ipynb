```python
import pandas as pd
import numpy as np
import torch
```

```python
compiled_csv = pd.read_csv("/media/solis/DATA/inference_emnist_letters/landscape_data_emnist_letters_last_best/resnet_emnist_letters/compiled_results.csv")
tensors = torch.load("/media/solis/DATA/inference_emnist_letters/landscape_data_emnist_letters_last_best/resnet_emnist_letters/Tensors/trainUval/a1_e48.pt")
```

```python
with open("bBR.txt") as f:
    bBR_idxs = list(map(int, f.read().split(",")))

with open("r.txt") as f:
    r_idxs = list(map(int, f.read().split(",")))
```

```python
Bb_idxs = compiled_csv.iloc[bBR_idxs]
Bb_idxs = Bb_idxs[Bb_idxs["Original_Label"] == 1]
Bb_idxs = list(Bb_idxs.index)
```

```python
with open("B.txt") as f:
    B_idxs = list(map(int, f.read().split(",")))

B_idxs = [idx for idx in B_idxs if idx in Bb_idxs]
b_idxs = [idx for idx in Bb_idxs if idx not in B_idxs]
```

```python
R_idxs = compiled_csv.iloc[bBR_idxs]
R_idxs = R_idxs[R_idxs["Original_Label"] == 17]
R_idxs = list(R_idxs.index)
```

```python
R_centroid = tensors[R_idxs].mean(dim=0)
Bb_centroid = tensors[Bb_idxs].mean(dim=0)
B_centroid = tensors[B_idxs].mean(dim=0)
b_centroid = tensors[b_idxs].mean(dim=0)
r_centroid = tensors[r_idxs].mean(dim=0)
```

```python
Rr_dist = torch.norm(R_centroid - r_centroid).item()
RBb_dist = torch.norm(R_centroid - Bb_centroid).item()
rBb_dist = torch.norm(r_centroid - Bb_centroid).item()
Bb_dist = torch.norm(B_centroid - b_centroid).item()
BR_dist = torch.norm(B_centroid - R_centroid).item()
Br_dist = torch.norm(B_centroid - r_centroid).item()
Rb_dist = torch.norm(R_centroid - b_centroid).item()
br_dist = torch.norm(b_centroid - r_centroid).item()
print(f"Distance between R and r centroids: {Rr_dist}")
print(f"Distance between R and Bb centroids: {RBb_dist}")
print(f"Distance between r and Bb centroids: {rBb_dist}")
print(f"Distance between B and b centroids: {Bb_dist}")
print(f"Distance between B and R centroids: {BR_dist}")
print(f"Distance between B and r centroids: {Br_dist}")
print(f"Distance between R and b centroids: {Rb_dist}")
print(f"Distance between b and r centroids: {br_dist}")
```
