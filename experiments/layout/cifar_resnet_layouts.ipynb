```python
%load_ext autoreload
%autoreload 2

import pyct as ct
from make_layout import make_layouts
from make_icicle import make_icicle

import pandas as pd

metrics = pd.read_csv("../../experiment_data/balance_metrics/cross_epoch_better.csv")

metrics = metrics[["dataset", "split", "model", "epoch", "layer", "train_acc", "val_acc"]]
metrics = metrics[(metrics["dataset"] == "cifar") & (metrics["model"] == "resnet")]
metrics = metrics[metrics["split"] == "val"]

metrics.sort_values(by=["epoch"], inplace=True)
metrics.reset_index(drop=True, inplace=True)
metrics.head()
```

```python
path_templ = lambda e: f"C:/home/sumatra_llvis_data/strees_cross_epoch_better/resnet_cifar/trainUval/ctree_a1_e{e}_20"
```

```python
for epoch in metrics["epoch"].unique():
	print(f"Making layout for epoch {epoch}")
	epoch_metrics = metrics[metrics["epoch"] == epoch]
	path = path_templ(epoch)
	# make_layouts(path, epoch_metrics)
	fig = make_icicle(path, style="remainder")
	fig.update_layout(margin=dict(t=30, b=10, l=10, r=10))
	train_acc = epoch_metrics["train_acc"].values[0]
	val_acc = epoch_metrics["val_acc"].values[0]
	fig.update_layout(title_text=f"ResNet CIFAR Train+Val Epoch {epoch} (Train: {train_acc}, Val: {val_acc})", title_x=0.5)
	fig.write_image(f"resnet_cifar_icicles_trainUval_remainder/e{epoch}.png", scale=4)
```
