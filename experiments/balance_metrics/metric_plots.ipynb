```python
%load_ext autoreload
%autoreload 2

import yaml
from balance_metrics import plot_epoch_metrics_vs_accuracies, get_data, plot_layer_metrics, plot_exp_regressions, plot_linear_regressions

epoch_path = "../../experiment_data/balance_metrics/cross_epoch.csv"
layer_path = "../../experiment_data/balance_metrics/cross_layer.csv"
block_path = "../../experiment_data/balance_metrics/intra_block.csv"

epoch_data = get_data(epoch_path)
layer_data = get_data(layer_path)
block_data = get_data(block_path)

with open('../layer_orders_cross_layer.yml', 'r') as f:
    layer_order = yaml.safe_load(f)

with open('../layer_orders_intra_block.yml', 'r') as f:
    block_order = yaml.safe_load(f)
```

```python
IGNORE_STDS = [0.0, 0.2, 0.5, 1.0, 2.0, 5.0]
suffix = lambda f: f"_std{f}" if f != 0.0 else ""
```

```python
for f in IGNORE_STDS:
	plot_epoch_metrics_vs_accuracies(epoch_path, '../../charts/balance/epoch_vs_accuracies' + suffix(f), ignore_std=f)
	plot_linear_regressions(epoch_data, '../../charts/balance/epoch_linear_regressions' + suffix(f), ignore_std=f)
	plot_exp_regressions(epoch_data, '../../charts/balance/epoch_exp_regressions' + suffix(f), ignore_std=f)
```

```python
for f in IGNORE_STDS:
	plot_layer_metrics(layer_path, '../../charts/balance/layer_metrics' + suffix(f), layer_order, ignore_std=f)
	plot_layer_metrics(block_path, '../../charts/balance/block_metrics' + suffix(f), block_order, ignore_std=f)
```
