```python
%load_ext autoreload
%autoreload 2

from load_mergetrees import load_mergetree_dir
from compute_edit_distances import compute_distance_matrix

paths = [f"/media/santripta/Santripta 1/strees_cross_epoch_{k}/resnet_cifar/trainUval" for k in [15, 20, 25, 30, 35, 40, 45, 60, 80]]

trees = [load_mergetree_dir(path, pattern=".*e143.*", simpl=1e-6) for path in paths]
treedict = {}
for td in trees:
    treedict.update(td)
```
