```python
import pickle as pkl

with open("cons_mat_ord.pkl", "rb") as f:
    mat_cons, ord_cons = pkl.load(f)

ks = [int(o.split("_")[-1]) for o in ord_cons]
ks
```

```python
import seaborn as sns
```

```python
fig = sns.heatmap(
	mat_cons[::-1],
	annot=True,
	fmt=".3f",
	cmap="BuPu",
	vmin=0,
	vmax=12.53174,
	cbar_kws={"label": "Distance"},
    xticklabels=ks,
    yticklabels=ks[::-1],
	linewidth=0.05,
	linecolor="#0000000a"
)

# fig.xaxis.tick_bottom()
fig = fig.set_ylabel("").get_figure()
fig.set_size_inches(7, 6)
fig.show()
fig.savefig(f"plots/resnet_distances_e143.png", bbox_inches="tight", dpi=300)
```
